```python
import json
from pathlib import Path
from urllib.parse import quote

import requests
from PIL import Image, ImageDraw


# Change this only if your folders have a different parent directory.
DATASET_ROOT = Path("/content/dataset")

TRAIN_APPLE = DATASET_ROOT / "train" / "apple"
TEST_APPLE = DATASET_ROOT / "test" / "apple"
TRAIN_NOT_APPLE = DATASET_ROOT / "train" / "not_apple"
TEST_NOT_APPLE = DATASET_ROOT / "test" / "not_apple"

for folder in [TRAIN_APPLE, TEST_APPLE, TRAIN_NOT_APPLE, TEST_NOT_APPLE]:
    folder.mkdir(parents=True, exist_ok=True)


IMAGE_SIZE = 256
PADDING = 13

CATEGORIES_URL = (
    "https://raw.githubusercontent.com/googlecreativelab/"
    "quickdraw-dataset/master/categories.txt"
)
RAW_DATA_URL = (
    "https://storage.googleapis.com/quickdraw_dataset/"
    "full/raw/{}.ndjson"
)


def safe_filename(value: str) -> str:
    return value.lower().replace(" ", "_").replace("/", "_")


def render_drawing(strokes: list) -> Image.Image | None:
    """Render Quick, Draw! strokes as a centred 256×256 grayscale PNG."""

    all_points = [
        (x, y)
        for stroke in strokes
        if len(stroke) >= 2
        for x, y in zip(stroke[0], stroke[1])
    ]

    if not all_points:
        return None

    min_x = min(x for x, _ in all_points)
    max_x = max(x for x, _ in all_points)
    min_y = min(y for _, y in all_points)
    max_y = max(y for _, y in all_points)

    drawing_width = max_x - min_x
    drawing_height = max_y - min_y
    largest_dimension = max(drawing_width, drawing_height)

    if largest_dimension == 0:
        return None

    scale = (IMAGE_SIZE - 2 * PADDING) / largest_dimension
    offset_x = (IMAGE_SIZE - drawing_width * scale) / 2
    offset_y = (IMAGE_SIZE - drawing_height * scale) / 2

    image = Image.new("L", (IMAGE_SIZE, IMAGE_SIZE), color=255)
    canvas = ImageDraw.Draw(image)

    for stroke in strokes:
        if len(stroke) < 2:
            continue

        x_points, y_points = stroke[0], stroke[1]

        points = [
            (
                (x - min_x) * scale + offset_x,
                (y - min_y) * scale + offset_y,
            )
            for x, y in zip(x_points, y_points)
        ]

        if len(points) >= 2:
            canvas.line(points, fill=0, width=2, joint="curve")
        elif len(points) == 1:
            x, y = points[0]
            canvas.ellipse((x - 1, y - 1, x + 1, y + 1), fill=0)

    return image


def export_category(
    category: str,
    training_folder: Path,
    testing_folder: Path,
    training_count: int,
    testing_count: int,
) -> None:
    """
    Stream a category from Google and create exactly the requested split.
    No NDJSON source file is saved to disk.
    """

    encoded_category = quote(category, safe="")
    source_url = RAW_DATA_URL.format(encoded_category)
    total_needed = training_count + testing_count
    saved = 0
    filename_prefix = safe_filename(category)

    print(f"Starting {category}...")

    with requests.get(source_url, stream=True, timeout=120) as response:
        response.raise_for_status()

        for line in response.iter_lines(decode_unicode=True):
            if saved >= total_needed:
                break

            if not line:
                continue

            record = json.loads(line)

            if not record.get("recognized", False):
                continue

            image = render_drawing(record["drawing"])

            if image is None:
                continue

            if saved < training_count:
                destination_folder = training_folder
                image_number = saved
            else:
                destination_folder = testing_folder
                image_number = saved - training_count

            image_path = (
                destination_folder
                / f"{filename_prefix}-{image_number:06d}.png"
            )

            image.save(image_path)
            saved += 1

            if saved % 1_000 == 0:
                print(f"  {category}: {saved:,} of {total_needed:,}")

    if saved != total_needed:
        raise RuntimeError(
            f"{category}: only found {saved:,} valid drawings; "
            f"needed {total_needed:,}."
        )

    print(f"Finished {category}: {saved:,} images")


# Download Google’s exact category list.
response = requests.get(CATEGORIES_URL, timeout=30)
response.raise_for_status()
categories = response.text.splitlines()

assert len(categories) == 345, f"Expected 345 categories, found {len(categories)}"

# Apple: 100,000 training + 20,000 test images.
export_category(
    category="apple",
    training_folder=TRAIN_APPLE,
    testing_folder=TEST_APPLE,
    training_count=100_000,
    testing_count=20_000,
)

# Every other category: 300 training + 60 test images.
for index, category in enumerate(categories, start=1):
    if category == "apple":
        continue

    print(f"\n[{index}/{len(categories)}]")
    export_category(
        category=category,
        training_folder=TRAIN_NOT_APPLE,
        testing_folder=TEST_NOT_APPLE,
        training_count=300,
        testing_count=60,
    )

print("\nDataset export complete.")
```
